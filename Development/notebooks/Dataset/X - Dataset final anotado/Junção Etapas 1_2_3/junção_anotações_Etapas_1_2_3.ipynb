```python
# Metrica Estatistica de Proporção de conflitos: % = conflitos/(conflitos + Nao_conflitos)
# Separação das etapas
etapas = {
    'Etapa 1': df_1,
    'Etapa 2': df_2,
    'Etapa 3': df_3
}

classes = {
    'Reply': 1,
    'Quote': 2,
    'Original': 0
}

# Calcula % de conflitos
dados = []

for etapa, df in etapas.items():

    linha = []

    for nome, codigo in classes.items():

        classe = df[df['Class'] == codigo]

        conflitos = (classe['Conflict'] == 1).sum()

        validos = (
            (classe['Conflict'] == 1).sum() +
            (classe['Conflict'] == 0).sum()
        )

        porcentagem = conflitos / validos * 100

        linha.append(porcentagem)

    dados.append(linha)

heat = pd.DataFrame(
    dados,
    index=['Etapa 1','Etapa 2','Etapa 3'],
    columns=['Reply','Quote','Original']
)

plt.figure(figsize=(6,4))

sns.heatmap(
    heat,
    annot=True,
    fmt=".1f",
    cmap="YlOrRd",
    vmin=0,
    vmax=100,
    cbar_kws={'label':'Conflitos (%)'}
)

plt.title("Percentual de pares classificados como conflito por classe de tweet e etapa de anotação")
plt.ylabel("")
plt.xlabel("Classe de tweet")

plt.tight_layout()
plt.show()
```
